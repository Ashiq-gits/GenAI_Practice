# 9. Handle missing values

Real time datasets often contain missing values.

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar", "Sneha", "Rahul"],
    "Age": [25, None, 30, 28, None],
    "Salary": [30000, 40000, None, 35000, 45000]
}

df = pd.DataFrame(data)

print(df)
```

# dropna() – Remove missing values

### It removes rows containing missing values by default.

axis=0 → Remove rows (default)
axis=1 → Remove columns.

```python
df_clean = df.dropna()

print(df_clean)
```

# In case if you can also remove columns containing missing values:

### Using axis=1

```python
df.dropna(axis=1)
```

# fillna() – Fill missing values

Instead of deleting rows, you can replace missing values.

Ex: Fill missing Age with 0

```python
df["Age"] = df["Age"].fillna(25)

print(df)
```

### Ex: Fill missing Salary with average salary

```python
df["Salary"] = df["Salary"].fillna(
    df["Salary"].mean()
)

print(df)
```

### Also fill missing text values:

```python
df["Name"] = df["Name"].fillna("Unknown")
print(df)
```

# remove() – Remove a row or column?

Important: Pandas DataFrames do not have a standard remove() method.

#### To remove a column, use drop():

```python
df.drop("Salary", axis=1, inplace=True)

print(df)
```

#### To remove a row using its index:

```python
df.drop(2, axis=0, inplace=True)

print(df)
```

axis=0 → Remove a row.
axis=1 → Remove a column.
inplace=True → Modify the original DataFrame.

# Comparison

Function	Purpose	Example
dropna()	Remove missing-value rows/columns	df.dropna()
fillna()	Fill missing values	df.fillna(0)
drop()	Remove specific rows/columns	df.drop("Age", axis=1)

# 1. Apply functions to DataFrame columns

The apply() function allows to execute custom logic on each row or column.

Ex: Categorize employee salaries.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Arun', 'Bala', 'Charan', 'David'],
    'Salary': [25000, 45000, 70000, 90000]
})

print(df)
print("------------------------------")

def salary_category(salary):
    if salary < 30000:
        return 'Low'
    elif salary < 60000:
        return 'Medium'
    else:
        return 'High'

df['Category'] = df['Salary'].apply(salary_category)

print(df)
```

# 2. Advanced filtering using isin(), between(), and query()

These functions are especially useful when working with large datasets.

```python
# Suppose we have employee data.

df = pd.DataFrame({
    'Name': ['Arun', 'Bala', 'Charan', 'David', 'Ezhil'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [30000, 45000, 70000, 90000, 50000]
})

print(df)
```

```python

df = pd.DataFrame({
    'Name': ['Arun', 'Bala', 'Charan', 'David', 'Ezhil'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [30000, 45000, 70000, 90000, 50000]
})

print(df)
print("---------------------------")

# Using isin():

result = df[df['Department'].isin(['IT', 'HR'])]

print(result)
```

```python
# Using between():

result = df[df['Salary'].between(40000, 80000)]

print(result)
```

```python
# Using query():

result = df.query("Salary > 40000 and Department == 'IT'")

print(result)
```

# 3. GroupBy with multiple aggregations


```python
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'HR', 'Finance'],
    'Salary': [50000, 60000, 30000, 40000, 70000],
    'Experience': [3, 5, 2, 4, 8]
})

print(df)
print("----------------------------")

result = df.groupby('Department').agg({
    'Salary': ['mean', 'max', 'min', 'sum'],
    'Experience': ['mean', 'max', 'min']
})

print(result)
```

# 4. Merge, Join and Concatenate
These operations combine multiple DataFrames.

# Types of joins:
    
###  | Join	| Description

#### inner ---->	Matching records only
#### left	----->  All records from left DataFrame
#### right ----->	All records from right DataFrame
#### outer ------>	All records from both DataFrame

```python
employees = pd.DataFrame({
    'EmpID': [1, 2, 3],
    'Name': ['Arun', 'Bala', 'Charan']
})

salary = pd.DataFrame({
    'EmpID': [1, 2, 3],
    'Salary': [30000, 40000, 50000]
})

result = pd.merge(
    employees,
    salary,
    on='EmpID',
    how='inner'
)

print(result)
```

```python
# Concat – Stack DataFrames

df1 = pd.DataFrame({'Name': ['Arun', 'Bala']})
df2 = pd.DataFrame({'Name': ['Charan', 'David']})

df3 = pd.concat([df1, df2])

print(df3)
```

# 5. Pivot Tables
Pivot tables summarize data in an Excel-like format.

It is useful for sales analysis, financial reports, and feature engineering.

```python

sales = pd.DataFrame({
    'Region': ['South', 'North', 'South', 'North'],
    'Product': ['Laptop', 'Laptop', 'Mobile', 'Mobile'],
    'Sales': [50000, 60000, 30000, 40000]
})

print(sales)
print("--------------------------")

pivot = pd.pivot_table(
    sales,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print(pivot)
```

# 6. Working with missing and duplicate data
Real datasets are rarely clean

```python
df = pd.DataFrame({
    'Name': ['Arun', 'Bala', 'Bala', 'David'],
    'Age': [25, None, None, 35],
    'Salary': [30000, 40000, 40000, None]
})

print(df)
```

```python
# Find missing values:

df.isnull().sum()
```

```python
# Fill missing values with median:

df['Age'] = df['Age'].fillna(df['Age'].median())

print(df)
```

```python
# Forward fill:

df['Salary'] = df['Salary'].ffill()

print(df)
```

```python
# Find duplicate rows:

# duplicated() – Identify duplicate rows

print(df.duplicated())

#print(df)
```

```python
# Remove duplicates:

df = df.drop_duplicates()

print(df)

# Important: Choose a missing value strategy based on the meaning of the column. 
# Forward fill is generally more appropriate for ordered time-series data than unrelated employee records.
```

# 7. String operations
Pandas provides the .str accessor for processing text columns.

```python
df = pd.DataFrame({
    'Name': [' arun ', 'BALA', 'charan'],
    'Email': ['arun@gmail.com',
              'bala@yahoo.com',
              'charan@gmail.com']
})

print(df)
```

```python
df['Name'] = df['Name'].str.strip() # Remove extra spaces - It removes unnecessary spaces from the beginning and end of text.
print(df)
```

```python
df['Name'] = df['Name'].str.title()
print(df)
```

```python
df['Email_Domain'] = df['Email'].str.split('@').str[1]
print(df)
```

```python
df['Is_Gmail'] = df['Email'].str.contains('gmail')
print(df)
```

```python
# Ex:

df['Name'] = df['Name'].str.strip() # Remove extra spaces - It removes unnecessary spaces from the beginning and end of text.

df['Name'] = df['Name'].str.title() # Convert text into title case - It converts the first letter of each word to uppercase and the remaining letters to lowercase.

df['Email_Domain'] = df['Email'].str.split('@').str[1] # Extract email domain

df['Is_Gmail'] = df['Email'].str.contains('gmail') # Check whether email contains Gmail What does it do?

print(df)
```

## Method	Purpose	Example
#### .str.upper()	All uppercase	ARUN
#### .str.lower()	All lowercase	arun
#### .str.title()	Title case	Arun Kumar
#### .str.strip()	Remove outside spaces	Arun
#### .str.replace()	Replace text	Change @gmail.com

# 8. Date and Time Operations
It is useful for stock market data, server logs, and time-series ML.

```python
df = pd.DataFrame({
    'Date': ['2026-01-01', '2026-01-02', '2026-01-03'],
    'Sales': [10000, 15000, 12000]
})

df['Date'] = pd.to_datetime(df['Date'])

print(df)
```

```python
# Extract year, month, and day:

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Day_Name'] = df['Date'].dt.day_name()

print(df)
```

```python
# Filter by date

result = df[
    df['Date'].between('2026-01-01', '2026-01-02')
]
print(result)
```
